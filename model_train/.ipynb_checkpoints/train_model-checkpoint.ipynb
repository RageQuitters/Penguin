```python
import numpy as np
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import LocalOutlierFactor
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, precision_recall_fscore_support

# =========================
# LOAD DATA
# =========================
df = pd.read_csv("dataset/machine_failure.csv")
df = df.drop(columns=["UDI", "Product ID", "Type"])

feature_cols = [c for c in df.columns if c not in
                ["Machine failure", "TWF", "HDF", "PWF", "OSF", "RNF"]]

targets = ["TWF", "HDF", "PWF", "OSF", "RNF"]

X = df[feature_cols]
y_global = df["Machine failure"]

# =========================
# TRAIN / TEST SPLIT
# =========================
X_train, X_test, y_train_global, y_test_global = train_test_split(
    X, y_global,
    test_size=0.2,
    random_state=42,
    stratify=y_global
)

# =========================
# SCALING
# =========================
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# =========================
# LOF
# =========================
normal_df = df[df["Machine failure"] == 0]
X_normal = scaler.fit_transform(normal_df[feature_cols])

lof = LocalOutlierFactor(
    n_neighbors=25,
    contamination=0.04,
    novelty=True
)
lof.fit(X_normal)

def anomaly_score(X):
    raw = lof.decision_function(X)
    return 1 / (1 + np.exp(-raw))

# =========================
# MODELS
# =========================
models = {}

print("\n================ TRAINING PHASE ================")

for t in targets:
    print(f"\nTraining {t}")

    y_train = df.loc[X_train.index, t]
    y_test = df.loc[X_test.index, t]

    model = RandomForestClassifier(
        n_estimators=200,
        max_depth=10,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_train_scaled, y_train)
    models[t] = model

    # =========================
    # 🔥 TRAIN METRICS
    # =========================
    train_pred = model.predict(X_train_scaled)
    print("\n--- TRAIN METRICS ---")
    print(classification_report(y_train, train_pred, zero_division=0))

    # =========================
    # 🔥 TEST METRICS
    # =========================
    test_pred = model.predict(X_test_scaled)
    print("\n--- TEST METRICS ---")
    print(classification_report(y_test, test_pred, zero_division=0))

# =========================
# MULTI-LABEL METRICS
# =========================
print("\n================ MULTI-LABEL METRICS ================")

y_true_all = []
y_pred_all = []

for t in targets:
    y_true_all.append(df.loc[X_test.index, t].values)
    y_pred_all.append(models[t].predict(X_test_scaled))

y_true_all = np.array(y_true_all).T
y_pred_all = np.array(y_pred_all).T

p, r, f1, _ = precision_recall_fscore_support(
    y_true_all,
    y_pred_all,
    average="weighted"
)

print("Precision:", p)
print("Recall   :", r)
print("F1 Score :", f1)

# =========================
# SYSTEM METRICS
# =========================
print("\n================ SYSTEM METRICS ================")

scores = anomaly_score(X_test_scaled)
THRESHOLD = np.percentile(scores, 92)

TP = TN = FP = FN = 0

for i in range(len(X_test_scaled)):

    score = scores[i]

    fv = [
        models[t].predict(X_test_scaled[i].reshape(1, -1))[0]
        for t in targets
    ]

    fault_count = sum(fv)
    true_label = y_test_global.iloc[i]

    if score > THRESHOLD and fault_count >= 2:
        pred = 1
    elif score > THRESHOLD or fault_count >= 1:
        pred = 1
    else:
        pred = 0

    if pred == 1 and true_label == 1:
        TP += 1
    elif pred == 0 and true_label == 0:
        TN += 1
    elif pred == 1 and true_label == 0:
        FP += 1
    else:
        FN += 1

precision = TP / (TP + FP + 1e-9)
recall = TP / (TP + FN + 1e-9)
f1 = 2 * precision * recall / (precision + recall + 1e-9)

print(f"TP: {TP}, TN: {TN}, FP: {FP}, FN: {FN}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")

# =========================
# PREDICT
# =========================
def predict(X_input):
    X_scaled = scaler.transform(X_input)
    score = anomaly_score(X_scaled)[0]

    fv = [models[t].predict(X_scaled)[0] for t in targets]
    fc = sum(fv)

    if score > THRESHOLD and fc >= 2:
        decision = "FAILURE"
    elif score > THRESHOLD or fc >= 1:
        decision = "WARNING"
    else:
        decision = "NORMAL"

    return {
        "anomaly_score": float(score),
        "failure_vector": fv,
        "fault_count": fc,
        "decision": decision
    }

# =========================
# SAMPLE
# =========================
sample = pd.DataFrame([{
    "Air temperature [K]": 298.2,
    "Process temperature [K]": 308.5,
    "Rotational speed [rpm]": 1400,
    "Torque [Nm]": 65.0,
    "Tool wear [min]": 191
}])

print("\n================ SAMPLE =================")
print(predict(sample))

# =========================
# SAVE
# =========================
joblib.dump(lof, "lof_model.joblib")
joblib.dump(models, "rf_models.joblib")
joblib.dump(scaler, "scaler.joblib")
joblib.dump(THRESHOLD, "threshold.joblib")

print("\nModels saved successfully!")
```
